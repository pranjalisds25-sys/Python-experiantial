```python
# ============================================================
# SMART HOME ENERGY CONSUMPTION MONITOR
# MODULE 1: APPLIANCE DATA COLLECTION
# ============================================================

import pandas as pd

print("=" * 60)
print("       SMART HOME ENERGY CONSUMPTION MONITOR")
print("              MODULE 1: DATA COLLECTION")
print("=" * 60)

# Ask the user how many appliances they want to enter
n = int(input("\nEnter the number of appliances: "))

# Empty lists to store the data
appliance_names = []
power_values = []
hours_values = []
days_values = []
rate_values = []

# Collect information for each appliance
for i in range(n):

    print("\n--------------------------------------------")
    print("Enter details for Appliance", i + 1)
    print("--------------------------------------------")

    name = input("Enter appliance name: ")

    power = float(input("Enter power consumption (Watts): "))

    hours = float(input("Enter usage hours per day: "))

    days = int(input("Enter number of days used per month: "))

    rate = float(input("Enter electricity rate (₹ per kWh): "))

    # Store the information
    appliance_names.append(name)
    power_values.append(power)
    hours_values.append(hours)
    days_values.append(days)
    rate_values.append(rate)


# Create a dictionary containing all collected data
data = {
    "Appliance": appliance_names,
    "Power_W": power_values,
    "Hours_Per_Day": hours_values,
    "Days_Per_Month": days_values,
    "Electricity_Rate": rate_values
}


# Convert the data into a Pandas DataFrame
df = pd.DataFrame(data)


# Display the collected information
print("\n\n" + "=" * 60)
print("              APPLIANCE DATA")
print("=" * 60)

display(df)


print("\n" + "=" * 60)
print("       MODULE 1 DATA COLLECTION COMPLETED")
print("=" * 60)
```

```python
# ============================================================
# SMART HOME ENERGY CONSUMPTION MONITOR
# MODULE 2: ENERGY CONSUMPTION & COST CALCULATION
# ============================================================

print("=" * 65)
print("       SMART HOME ENERGY CONSUMPTION MONITOR")
print("          MODULE 2: ENERGY & COST CALCULATION")
print("=" * 65)


# ------------------------------------------------------------
# CHECK WHETHER MODULE 1 DATA IS AVAILABLE
# ------------------------------------------------------------

if "df" not in globals():
    print("\nERROR: Module 1 data was not found.")
    print("Please run Module 1 first.")
else:

    # --------------------------------------------------------
    # CALCULATE MONTHLY ENERGY CONSUMPTION
    # Formula:
    # Energy = Power × Hours per Day × Days per Month / 1000
    # --------------------------------------------------------

    df["Energy_kWh"] = (
        df["Power_W"]
        * df["Hours_Per_Day"]
        * df["Days_Per_Month"]
    ) / 1000


    # --------------------------------------------------------
    # CALCULATE MONTHLY ELECTRICITY COST
    # Formula:
    # Cost = Energy × Electricity Rate
    # --------------------------------------------------------

    df["Monthly_Cost"] = (
        df["Energy_kWh"]
        * df["Electricity_Rate"]
    )


    # --------------------------------------------------------
    # DISPLAY APPLIANCE-WISE RESULTS
    # --------------------------------------------------------

    print("\n")
    print("=" * 65)
    print("          APPLIANCE-WISE ENERGY & COST")
    print("=" * 65)

    display(df)


    # --------------------------------------------------------
    # CALCULATE TOTAL ENERGY CONSUMPTION
    # --------------------------------------------------------

    total_energy = df["Energy_kWh"].sum()


    # --------------------------------------------------------
    # CALCULATE TOTAL ELECTRICITY COST
    # --------------------------------------------------------

    total_cost = df["Monthly_Cost"].sum()


    # --------------------------------------------------------
    # FIND THE APPLIANCE WITH HIGHEST ENERGY CONSUMPTION
    # --------------------------------------------------------

    highest_energy_index = df["Energy_kWh"].idxmax()

    highest_appliance = df.loc[
        highest_energy_index, "Appliance"
    ]

    highest_energy = df.loc[
        highest_energy_index, "Energy_kWh"
    ]


    # --------------------------------------------------------
    # FIND THE APPLIANCE WITH HIGHEST ELECTRICITY COST
    # --------------------------------------------------------

    highest_cost_index = df["Monthly_Cost"].idxmax()

    highest_cost_appliance = df.loc[
        highest_cost_index, "Appliance"
    ]

    highest_cost = df.loc[
        highest_cost_index, "Monthly_Cost"
    ]


    # --------------------------------------------------------
    # DISPLAY MONTHLY SUMMARY
    # --------------------------------------------------------

    print("\n")
    print("=" * 65)
    print("                  MONTHLY SUMMARY")
    print("=" * 65)

    print(f"Total Energy Consumption : {total_energy:.2f} kWh")
    print(f"Total Electricity Cost   : ₹{total_cost:.2f}")

    print("\nHighest Energy Consuming Appliance:")
    print(f"{highest_appliance} - {highest_energy:.2f} kWh")

    print("\nHighest Electricity Cost Appliance:")
    print(f"{highest_cost_appliance} - ₹{highest_cost:.2f}")

    print("=" * 65)
    print("       MODULE 2 COMPLETED SUCCESSFULLY")
    print("=" * 65)
```
